**Chosen API:** PokéAPI, which contains data about Pokemon and the Pokemon games.
Each Pokemon character in the games has specific statistics that we will retrieve and display using Pandas

**Goal:**
1. Request data from the API about the first 10 Pokémon.
2. Retrieve their types, attack, defense, speed, height, and weight.
3. Use the pandas module to organize and display the data in a table.

### Step 1: Requesting data from PokéAPI

PokéAPI is a public API that provides information about Pokemon, including their types, stats, and other characteristics.

I will retrieve the first 10 Pokémon listed.

```python

#requests library to request from API
import requests

#url from PokeAPI containing the first 10 pokemon
url_pokemon = "https://pokeapi.co/api/v2/pokemon?limit=10&offset=0"

response = requests.get(url_pokemon, timeout=15)

#checks whether the request was successful or returns an error
response.raise_for_status()

#converts the response into a Python dictionary
data_10_pokemon = response.json()


#prints the names of the first 10 pokemon
for pokemon in data_10_pokemon["results"]:
    print(pokemon["name"])
```

### Step 2: Retrieving individual Pokemon data

The initial API request returns a list of Pokémon names and URLs, rather than all their individual characteristics.

Each Pokemon has its own URL containing more detailed information. I will use these URLs to request the types, stats, height, and weight of each Pokemon.

The code also checks whether each request was successful before processing the response.

```python

#empty list that will store info on each pokemon
pokemon_data = []

#loops through the first 10 pokemon from the API response
for pokemon in data_10_pokemon["results"]:

    #requests detailed data for the current pokemon using their own url
    pokemon_url = pokemon["url"]
    pokemon_response = requests.get(pokemon_url, timeout=15)
    pokemon_response.raise_for_status()

    #converts the response into a Python dictionary
    pokemon_details = pokemon_response.json()

    #empty list to store pokemon types
    pokemon_types = []
    for pokemon_type in pokemon_details["types"]:
        pokemon_types.append(pokemon_type["type"]["name"])

    #empty dictionary to store pokemon statistics
    stats = {}
    for stat in pokemon_details["stats"]:
        stats[stat["stat"]["name"]] = stat["base_stat"]

    #stores the specific info we want: type, atk, def, spd, wgt, hgt
    pokemon_data.append({
        "pokemon": pokemon_details["name"],
        "type": ", ".join(pokemon_types),
        "attack": stats["attack"],
        "defense": stats["defense"],
        "speed": stats["speed"],
        "height": pokemon_details["height"],
        "weight": pokemon_details["weight"]
    })

#displays the collected data
pokemon_data
```

### Step 3: Organizing the data with pandas

The information collected from the API is stored in a list of dictionaries. I will use the pandas module to convert this information into a DataFrame, which organizes the data more cleanly

I will also convert height from decimeters to meters and weight from hectograms to kilograms, making the measurements easier to interpret.

```python

#imports pandas to organize data into a table
import pandas as pd

#converts the pokemon data into a pandas dataframe
pokemon_df = pd.DataFrame(pokemon_data)

#converts height from decimeters to meters
pokemon_df["height"] = pokemon_df["height"] / 10

#converts weight from hectograms to kilograms
pokemon_df["weight"] = pokemon_df["weight"] / 10

#renames columns to make units clear
pokemon_df = pokemon_df.rename(columns={
    "height": "height (m)",
    "weight": "weight (kg)"
})

#displays the pokemon data as a table
pokemon_df
```

### Results

The resulting DataFrame summarizes the types, attack, defense, speed, height, and weight of the first 10 Pokémon listed in PokéAPI. 

Using pandas makes it easier to compare these characteristics in a structured table.
